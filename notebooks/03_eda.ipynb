```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the cleaned dataset
file_path = r"C:\Users\sushm\OneDrive\Desktop\Digital-Wellbeing-Analytics\data\processed\smartphone_usage_cleaned.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("Addiction label counts:")
print(df["addicted_label"].value_counts())

print("\nAddiction label percentages:")
print(df["addicted_label"].value_counts(normalize=True) * 100)
```

```python
df["addicted_label"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(6, 4)
)

plt.title("Distribution of Smartphone Addiction")
plt.xlabel("Addicted Label (0 = Not Addicted, 1 = Addicted)")
plt.ylabel("Number of Users")
plt.xticks(rotation=0)
plt.show()
```

```python
print("Average daily screen time by addiction status:")

print(
    df.groupby("addicted_label")["daily_screen_time_hours"].mean()
)
```

```python
screen_time_avg = df.groupby("addicted_label")["daily_screen_time_hours"].mean()

plt.figure(figsize=(7, 5))

plt.bar(
    ["Not Addicted", "Addicted"],
    screen_time_avg.values
)

plt.title("Average Daily Screen Time by Addiction Status")
plt.xlabel("Addiction Status")
plt.ylabel("Average Screen Time (Hours)")
plt.show()
```

```python
stress_addiction = pd.crosstab(
    df["stress_level"],
    df["addicted_label"],
    normalize="index"
) * 100

print("Addiction percentage within each stress level:")
print(stress_addiction)
```

```python
stress_addiction.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Addiction Distribution Across Stress Levels")
plt.xlabel("Stress Level")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.legend(["Not Addicted", "Addicted"])
plt.show()
```

```python
import pandas as pd

file_path = r"C:\Users\sushm\OneDrive\Desktop\Digital-Wellbeing-Analytics\data\processed\smartphone_usage_cleaned.csv"

df = pd.read_csv(file_path)

df.shape
```

```python
df["total_leisure_hours"] = (
    df["social_media_hours"] + df["gaming_hours"]
)

df[["social_media_hours", "gaming_hours", "total_leisure_hours"]].head()
```

```python
df["screen_to_sleep_ratio"] = (
    df["daily_screen_time_hours"] / df["sleep_hours"]
)

df[[
    "daily_screen_time_hours",
    "sleep_hours",
    "screen_to_sleep_ratio"
]].head()
```

```python
df["weekend_screen_time_increase"] = (
    df["weekend_screen_time"] - df["daily_screen_time_hours"]
)

df[[
    "daily_screen_time_hours",
    "weekend_screen_time",
    "weekend_screen_time_increase"
]].head()
```

```python
df["notification_intensity"] = (
    df["notifications_per_day"] / df["app_opens_per_day"]
)

df[[
    "notifications_per_day",
    "app_opens_per_day",
    "notification_intensity"
]].head()
```

```python
df["productive_time_ratio"] = (
    df["work_study_hours"] /
    (df["work_study_hours"] + df["daily_screen_time_hours"])
)

df[[
    "work_study_hours",
    "daily_screen_time_hours",
    "productive_time_ratio"
]].head()
```

```python
import numpy as np

# Check missing values in the five new features
new_features = [
    "total_leisure_hours",
    "screen_to_sleep_ratio",
    "weekend_screen_time_increase",
    "notification_intensity",
    "productive_time_ratio"
]

print("Missing values:")
print(df[new_features].isnull().sum())

print("\nInfinite values:")
print(np.isinf(df[new_features]).sum())
```
